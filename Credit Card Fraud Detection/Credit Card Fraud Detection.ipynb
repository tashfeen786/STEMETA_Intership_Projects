```python
# Apply Normalizition on 'Amount' and 'Time'
scaler = StandardScaler()
df['Amount'] = scaler.fit_transform(df[['Amount']])
df['Time'] = scaler.fit_transform(df[['Time']])
```

```python
sns.countplot(x='Class', data=df)
plt.title("Before Balancing")
plt.show()
```

```python
# Split into minority and majority classes
fraud = df[df['Class'] == 1]
non_fraud = df[df['Class'] == 0]
```

```python
# Apply Undersampling to majority class
from sklearn.utils import resample

non_fraud_downsampled = resample(non_fraud,
                                 replace=False,
                                 n_samples=len(fraud),
                                 random_state=42)
```

```python

# Combine both
balanced_df = pd.concat([non_fraud_downsampled, fraud])

print("\nAfter balancing:")
print(balanced_df['Class'].value_counts())
```

```python
sns.countplot(x='Class', data=balanced_df)
plt.title("After Balancing")
plt.show()
```

```python
# Feature / Target Split
X = balanced_df.drop('Class', axis=1)
y = balanced_df['Class']
```

```python
# Split Data into Training and Testing Sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Model Training
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
# Predictions
y_pred = model.predict(X_test)
```

```python

# Evaluation
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Train predictions
y_train_pred = model.predict(X_train)

# Test predictions
y_test_pred = model.predict(X_test)

# Function to compute metrics
def get_scores(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred, zero_division=0),
        "Recall": recall_score(y_true, y_pred, zero_division=0),
        "F1-score": f1_score(y_true, y_pred, zero_division=0)
    }

# Get scores
train_scores = get_scores(y_train, y_train_pred)
test_scores = get_scores(y_test, y_test_pred)
```

```python
import matplotlib.pyplot as plt

# Metrics to plot
metrics = list(train_scores.keys())
train_values = list(train_scores.values())
test_values = list(test_scores.values())

x = range(len(metrics))

plt.figure(figsize=(10,6))
plt.bar(x, train_values, width=0.4, label='Train', align='center')
plt.bar([p + 0.4 for p in x], test_values, width=0.4, label='Test', align='center')

plt.xticks([p + 0.2 for p in x], metrics)
plt.ylabel('Score')
plt.ylim(0, 1)
plt.title('Model Performance Comparison: Train vs Test')
plt.legend()
plt.grid(axis='y')
plt.show()
```
